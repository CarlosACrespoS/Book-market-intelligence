### 1.1 Análisis de Vigencia: El "Timestamp" de la Competencia
¿Qué tan moderno es el servicio que queremos superar? Evaluamos el volumen de su catálogo publicado a partir del **1 de enero de 2000**. 

**Por qué este dato es oro para nuestra Propuesta de Valor:**
* **Detección de "Inventario Zombie":** Si el competidor tiene muchos libros pero pocos son modernos (post-2000), nuestra startup puede diferenciarse posicionándose como la "plataforma de las últimas tendencias".
* **Afinidad con el Usuario Digital:** Los usuarios que adoptaron nuevas apps en la pandemia buscan contenido fresco. Saber cuántos títulos recientes ofrece la competencia nos permite ajustar nuestra estrategia de adquisición de licencias para superarlos en relevancia.

**¿Por qué es vital este Insight para la Startup?**
* **Posicionamiento Competitivo:** Determina si nuestra propuesta de valor se inclina hacia el coleccionismo clásico o hacia las tendencias actuales de consumo digital.
* **Segmentación de Audiencia:** Un catálogo joven atrae a usuarios más activos en plataformas tecnológicas, quienes fueron los principales adoptantes de nuevas apps durante el confinamiento.

```python
# Consulta para comparar el universo total vs. contenido reciente
query_universe = """
SELECT 
    COUNT(book_id) AS total_books,
    COUNT(CASE WHEN publication_date > '2000-01-01' THEN 1 END) AS recent_books,
    ROUND(COUNT(CASE WHEN publication_date >= '2000-01-01' THEN 1 END) * 100.0 / COUNT(book_id), 2) AS percentage_recent
FROM books;
"""

df_universe = pd.io.sql.read_sql(query_universe, con=engine)
print(f"Hallazgo Estratégico: El competidor ofrece {df_universe.iloc[0,1]} títulos de la era moderna.")
display(df_universe)
```

## 💎 Pilar 2: Diagnóstico de Calidad y Popularidad
*(Identificación de títulos "ancla" mediante métricas cuantitativas y cualitativas)*


Tras validar la vigencia del catálogo, el **desarrollo de la propuesta** exige una disección del rendimiento de los activos del competidor. No basta con saber qué libros ofrecen, sino cuáles de ellos son los "motores reales" de su plataforma y cuáles son simplemente ruido en el inventario.

### 2.1 Auditoría de Activos: Popularidad Masiva vs. Calidad Real

Para el **desarrollo**, hemos diseñado una métrica de auditoría que desglosa la interacción del servicio competidor en dos dimensiones críticas: **Cuantitativa (Calificaciones)** y **Cualitativa (Reseñas de Texto)**. 

**¿Por qué es vital esta distinción para nuestra estrategia de mercado?**
En el entorno digital, un libro puede recibir cientos de "clics rápidos" (estrellas) con poco esfuerzo, pero solo los títulos que provocan reseñas escritas generan una comunidad leal. Al contrastar el promedio de satisfacción con el volumen de participación, medimos la **solidez del prestigio** del rival:

* **Validación de Riesgo:** Identificar promedios de calificación "frágiles" (pocos votos) nos permite evitar errores que el competidor está cometiendo y atacar con títulos mejor validados estadísticamente.
* **Detección de "Engagement" Profundo:** Los libros con alta densidad de reseñas de texto son los activos que realmente fidelizan al usuario post-pandemia. Nuestra propuesta debe capturar este nivel de conversación activa para superar la frialdad del sistema del rival.

Esta auditoría nos permite mapear qué libros del competidor son "ruido" y cuáles son los verdaderos pilares de su éxito que debemos emular o superar.

```python
# Consulta SQL para traer la totalidad del catálogo con sus métricas agregadas
query_metrics_full = """
WITH book_ratings AS (
    SELECT 
        book_id,
        COUNT(rating_id) AS total_ratings,
        ROUND(AVG(rating), 2) AS avg_rating,
        MIN(rating) AS min_rating,
        MAX(rating) AS max_rating
    FROM ratings
    GROUP BY book_id
),
book_reviews AS (
    SELECT 
        book_id,
        COUNT(review_id) AS total_reviews
    FROM reviews
    GROUP BY book_id
)
SELECT 
    b.book_id,
    b.title AS "Título",
    COALESCE(r.total_ratings, 0) AS "Total Calificaciones",
    COALESCE(r.avg_rating, 0.00) AS "Calificación Promedio",
    COALESCE(r.min_rating, 0) AS "Calificación Mínima",
    COALESCE(r.max_rating, 0) AS "Calificación Máxima",
    COALESCE(v.total_reviews, 0) AS "Total Reseñas de Texto"
FROM books AS b
LEFT JOIN book_ratings AS r ON b.book_id = r.book_id
LEFT JOIN book_reviews AS v ON b.book_id = v.book_id
ORDER BY "Total Calificaciones" DESC;
"""

# Carga de la totalidad de los datos
df_books_full = pd.io.sql.read_sql(query_metrics_full, con=engine)

# Presentación del Dataset Completo
print(f"Dimensiones del Dataset Completo: {df_books_full.shape[0]} filas × {df_books_full.shape[1]} columnas")
print("Muestra general del inventario completo:")

# Al desplegar df_books_full directamente (o ajustando pd.set_option si lo deseas), 
# se visualiza la estructura global del catálogo completo.
display(df_books_full)
```

## 🏭 Pilar 3: Inteligencia de Suministro (Publishers)
*(Evaluación de eficiencia editorial y engagement por proveedor)*

Para el **desarrollo de la propuesta**, es fundamental identificar a los socios comerciales que sostienen la infraestructura de contenido del competidor. No todos los proveedores aportan el mismo valor; por ello, auditamos quiénes dominan el segmento de "lectura profunda" (obras > 50 páginas), descartando folletos o material efímero que no genera retención a largo plazo.

### 3.1 Identificación de Proveedores de Alto Volumen

Este análisis nos permite mapear el poder de negociación y la dependencia que el servicio competidor tiene con las grandes casas editoriales. Al identificar a los líderes en volumen de contenido robusto, podemos trazar nuestra propia estrategia de adquisición de derechos.

**Objetivos estratégicos para el desarrollo:**
* **Mapeo de Alianzas:** ¿Con quién tiene el competidor sus contratos más fuertes? 
* **Detección de Barreras de Entrada:** Si una editorial domina el 40% del catálogo del rival, nuestra propuesta debe buscar alternativas o negociar con competidores de ese proveedor para diferenciar la oferta.
* **Calidad del Inventario:** Filtrar por más de 50 páginas asegura que estamos analizando activos editoriales reales y no contenido promocional.

```python
# Consulta para encontrar editoriales con más de 50 páginas en sus libros
query_publishers = """
SELECT 
    p.publisher,
    COUNT(b.book_id) AS total_books
FROM publishers AS p
JOIN books AS b 
    ON p.publisher_id = b.publisher_id
WHERE b.num_pages > 50
GROUP BY p.publisher
ORDER BY total_books DESC
LIMIT 10;
"""

df_publishers = pd.io.sql.read_sql(query_publishers, con=engine)
display(df_publishers)
```

### 3.2 Análisis de Eficiencia Editorial: Volumen vs. Engagement

Para el **desarrollo de la propuesta**, no basta con conocer el tamaño del inventario del competidor. Debemos descubrir la relación real entre su presencia en el catálogo (*Market Share*) y la tracción que generan sus títulos (*Engagement*). 

En esta fase, calculamos el **Engagement Ratio**, que promedia las reseñas y calificaciones obtenidas por cada libro publicado bajo cada sello. Este KPI es fundamental para distinguir entre:
1. **Proveedores de Volumen:** Generan "bulto" en el catálogo pero con interacción moderada.
2. **Proveedores de Impacto:** Sellos que, independientemente de su tamaño, mantienen una comunidad de lectores extremadamente activa.

**¿Por qué es crítico para el desarrollo?**
Identificar dónde el competidor tiene mucho volumen pero poco impacto (bajo ratio) nos revela activos ineficientes. Por el contrario, los altos ratios señalan editoriales que nuestra propuesta debe priorizar para garantizar una base de usuarios vibrante desde el primer día.

```python
# Consulta SQL del Pilar/Análisis de Editoriales
query_publisher_ratio = """
WITH publisher_metrics AS (
    SELECT 
        b.publisher_id,
        p.publisher AS editorial,
        COUNT(DISTINCT b.book_id) AS total_libros,
        COUNT(r.rating_id) AS total_calificaciones,
        COUNT(v.review_id) AS total_reseñas
    FROM books AS b
    LEFT JOIN publishers AS p ON b.publisher_id = p.publisher_id
    LEFT JOIN ratings AS r ON b.book_id = r.book_id
    LEFT JOIN reviews AS v ON b.book_id = v.book_id
    GROUP BY b.publisher_id, p.publisher
    -- Filtro de volumen mínimo para evitar sesgos con editoriales de 1 solo libro (opcional según el criterio del proyecto)
    HAVING COUNT(DISTINCT b.book_id) >= 5 
)
SELECT 
    editorial AS "Editorial",
    total_libros AS "Total Libros",
    total_calificaciones AS "Total Calificaciones",
    total_reseñas AS "Total Reseñas",
    -- Cálculo explícito del ratio de calificaciones por libro
    ROUND(total_calificaciones::numeric / total_libros, 2) AS "Ratio Calificaciones/Libro",
    -- Cálculo del ratio de reseñas por libro
    ROUND(total_reseñas::numeric / total_libros, 2) AS "Ratio Reseñas/Libro"
FROM publisher_metrics
-- CORRECCIÓN CLAVE: Ordenar por la métrica de ratio/eficiencia y NO por el volumen de libros
ORDER BY "Ratio Calificaciones/Libro" DESC;
"""

df_publisher_ratio = pd.io.sql.read_sql(query_publisher_ratio, con=engine)

print("--- ANÁLISIS DE EFICIENCIA POR EDITORIAL (ORDENADO POR RATIO) ---")
display(df_publisher_ratio.head(10))
```

## ✒️ Pilar 4: Prestigio de Autor y Blindaje de Marca
*(Validación de reputación estadística para la reducción de riesgo)*

Para el **desarrollo de la propuesta**, identificar a las figuras que sostienen la credibilidad del competidor es un paso de "blindaje". En esta fase, buscamos detectar a los autores que no solo tienen fama, sino una **reputación estadística irrefutable**. 

### 4.1 Identificación de Activos de Autoría (Mínimo 50 Calificaciones)

En este análisis, respondemos a una pregunta clave: *¿Quiénes son los autores del competidor con los libros mejor calificados bajo una muestra representativa?* Para evitar sesgos por promedios inflados con pocos votos, aplicamos un **Filtro de Relevancia de 50 calificaciones**.

**Valor estratégico para el desarrollo:**
* **Garantía de Satisfacción:** Identificamos las obras que el mercado ya ha validado como "excelentes".
* **Reducción de Riesgo:** Seleccionar autores con estos indicadores para nuestra propuesta minimiza el riesgo de rechazo del usuario.
* **Benchmark de Calidad:** Establecemos el estándar de nota (ej. 4.41) que nuestro contenido estrella deberá alcanzar para ser competitivo.

```python
# Autor con mayor calificación promedio
query_top_author = """
WITH valid_books AS (
    -- Filtrando libros con AL MENOS 50 calificaciones (>= 50)
    SELECT 
        book_id,
        COUNT(rating_id) AS total_ratings
    FROM ratings
    GROUP BY book_id
    HAVING COUNT(rating_id) >= 50
)
-- 2. Agrupando por autor considerando únicamente sus libros válidos
SELECT 
    a.author AS "Autor",
    COUNT(DISTINCT b.book_id) AS "Libros Calificados (>=50)",
    SUM(vb.total_ratings) AS "Total Calificaciones",
    ROUND(AVG(r.rating), 2) AS "Calificación Promedio Autor"
FROM valid_books AS vb
JOIN books AS b ON vb.book_id = b.book_id
JOIN authors AS a ON b.author_id = a.author_id
JOIN ratings AS r ON b.book_id = r.book_id
GROUP BY a.author_id, a.author
ORDER BY "Calificación Promedio Autor" DESC
LIMIT 10;
"""

df_top_author = pd.io.sql.read_sql(query_top_author, con=engine)

print("--- TOP AUTORES CON MAYOR CALIFICACIÓN PROMEDIO (LIBROS CON >= 50 CALIFICACIONES) ---")
display(df_top_author)
```

## 👥 Pilar 5: El Factor Humano y Capital Social
*(Análisis de Power Users y la solidez de la prueba social)*

Para finalizar la auditoría, el **desarrollo de la propuesta** debe centrarse en el activo más difícil de replicar: la comunidad. No analizamos usuarios de forma genérica, sino que aplicamos un **Análisis de Embudo** para identificar a los "evangelizadores" del servicio competidor.

### 5.1 Análisis de Segmentación: Del Usuario Casual al Power User

Este estudio nos permite medir la salud social del rival. Una plataforma puede tener miles de usuarios, pero su verdadera fuerza reside en los **Power Users**: aquellos cuya lealtad y actividad sostienen la prueba social (reseñas) que convence a otros de leer.

**Dimensiones críticas para el desarrollo:**
1. **Alcance del Mercado:** Total de usuarios únicos que interactúan con el rival.
2. **Identificación de la Élite (Power Users):** Usuarios con una actividad extrema (> 50 calificaciones).
3. **Productividad Cualitativa:** Cuántas reseñas de texto aporta este segmento. Si este número es alto, el competidor tiene una barrera de entrada fuerte debido al contenido generado por el usuario (UGC).

```python
# Consulta de Embudo: Total Usuarios vs Power Users vs Reseñas Power Users vs Total Universo Reseñas
query_user_funnel = """
WITH power_users AS (
    -- 1. Identificar usuarios con más de 50 calificaciones
    SELECT username
    FROM ratings
    GROUP BY username
    HAVING COUNT(book_id) > 50
),
power_user_reviews AS (
    -- 2. Contar las reseñas de texto escritas por los Power Users
    SELECT 
        username,
        COUNT(review_id) AS review_count
    FROM reviews
    WHERE username IN (SELECT username FROM power_users)
    GROUP BY username
)
SELECT 
    (SELECT COUNT(DISTINCT username) FROM ratings) AS total_users_platform,
    (SELECT COUNT(*) FROM power_users) AS count_power_users,
    ROUND(AVG(review_count), 2) AS avg_text_reviews_power_users,
    -- Novedad: Suma total de reseñas creadas por Power Users
    COALESCE(SUM(review_count), 0) AS total_reviews_power_users,
    -- Novedad: Universo total de reseñas de texto en toda la plataforma
    (SELECT COUNT(*) FROM reviews) AS total_reviews_universe
FROM power_user_reviews;
"""

df_funnel = pd.io.sql.read_sql(query_user_funnel, con=engine)
display(df_funnel)
```

### 📊 Interpretación del Capital Social: Diagnóstico de la Comunidad

Los resultados del embudo de usuarios del competidor aportan datos clave para entender la dinámica de su comunidad y fundamentar el **desarrollo de nuestra propuesta**:

* **Proporción de Usuarios Frecuentes:** De un universo de **160 usuarios**, únicamente **6 se identifican como Power Users** (más de 50 calificaciones), lo que representa el **3.75%** del total de la comunidad.
* **Intensidad Individual de Escritura:** Estos 6 usuarios presentan una alta fidelidad y un hábito constante, promediando **24.33 reseñas de texto** por persona (con un total acumulado de **146 reseñas**).
* **Distribución Real del Contenido:** Frente a un universo global de **2,793 reseñas de texto**, las aportaciones de este grupo representan el **5.23%** del contenido escrito total de la plataforma. Esto demuestra que la comunidad del competidor **no depende de un monopolio de pocos creadores**, sino que la generación de contenido se encuentra ampliamente distribuida entre el 96.25% restante de los usuarios.

### 5.2 Identificación de "Super-Contribuidores": Perfil de Usuarios Frecuentes

**Objetivos analíticos para la propuesta:**

* **Perfil de Usuario Promedio vs. Frecuente:** Comprender las diferencias de comportamiento entre la comunidad general y los usuarios frecuentes para diseñar incentivos de retención más eficientes en nuestra plataforma.

```python
# Consulta para identificar a los 6 Power Users y su volumen de reseñas
query_top_6_users = """
SELECT 
    username AS "Usuario",
    COUNT(review_id) AS "Total Reseñas de Texto"
FROM reviews
WHERE 
    username IN (
        SELECT username
        FROM ratings
        GROUP BY username
        HAVING COUNT(book_id) > 50
    )
GROUP BY username
ORDER BY "Total Reseñas de Texto" DESC;
"""

df_top_6 = pd.io.sql.read_sql(query_top_6_users, con=engine)
display(df_top_6)
```
